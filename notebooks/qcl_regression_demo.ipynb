# Quantum Circuit Learning Regression

This notebook walks through training a small quantum circuit to learn a smooth function and compares it with a classical regressor.

```python
%config InlineBackend.figure_formats = ['svg']
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

# Make sure the app package is importable when running the notebook
repo_root = Path('..').resolve()
sys.path.append(str(repo_root))

from app.dataset import generate_function_data, split_data
from app.trainer import train_qcl, evaluate_qcl
from app.classical_baseline import train_classical_regressor, evaluate_classical_regressor
from app.plots import plot_qcl_fit, plot_classical_fit, plot_training_loss
```

```python
np.random.seed(0)

# Generate a small dataset for the demo
x, y = generate_function_data(n_samples=80, noise=0.02)
print(f'Shape of x: {x.shape}, shape of y: {y.shape}')

x_train, x_val, x_test, y_train, y_val, y_test = split_data(x, y)
```

```python
# Train the quantum model with a few epochs to keep the demo quick
weights, history = train_qcl(x_train, y_train, x_val, y_val, n_epochs=40, lr=0.15, n_layers=2)

qcl_mse, qcl_r2, qcl_preds = evaluate_qcl(x_test, y_test, weights)
print(f'QCL test MSE: {qcl_mse:.4f}, R^2: {qcl_r2:.4f}')
```

```python
# Train a classical baseline for comparison
classical_model = train_classical_regressor(x_train, y_train)
classical_mse, classical_r2, classical_preds = evaluate_classical_regressor(classical_model, x_test, y_test)
print(f'Classical test MSE: {classical_mse:.4f}, R^2: {classical_r2:.4f}')
```

```python
# Plot the fits and loss curves inline (SVG).
plot_qcl_fit(x_test, y_test, qcl_preds, 'examples/qcl_function_fit.svg')
plot_classical_fit(x_test, y_test, classical_preds, 'examples/classical_function_fit.svg')
plot_training_loss(history, 'examples/qcl_training_loss.svg')

from IPython.display import SVG, display
for svg_path in ['examples/qcl_function_fit.svg', 'examples/classical_function_fit.svg', 'examples/qcl_training_loss.svg']:
    display(SVG(filename=svg_path))
```
